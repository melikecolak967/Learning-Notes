```python
# 2-PDF İşleme Fonksiyonları

import PyPDF2
from langchain_text_splitters import RecursiveCharacterTextSplitter, SentenceTransformersTokenTextSplitter

# PDF'den metin çıkarma
def convert_PDF_Text(pdf_path):
    try:
        reader = PyPDF2.PdfReader(pdf_path)
        pdf_pages = [p.extract_text().strip() for p in reader.pages]
        pdf_pages = [text for text in pdf_pages if text]
        print(f"✅ PDF okundu: {len(pdf_pages)} sayfa")
        return pdf_pages
    except Exception as e:
        print(f"❌ Hata: {e}")
        return []

# Metni karakterlere göre parçalama
def convert_Page_ChunkinChar(pdf_texts, chunk_size=1500, chunk_overlap=200):
    character_splitter = RecursiveCharacterTextSplitter(
        separators=["\n\n", "\n", ". ", " ", ""],
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap
    )
    character_split_texts = character_splitter.split_text('\n\n'.join(pdf_texts))
    print(f"✅ {len(character_split_texts)} karakter parçası oluşturuldu")
    return character_split_texts

# Token bazlı parçalama
def convert_Chunk_Token(text_chunksinChar, sentence_transformer_model, chunk_overlap=10, tokens_per_chunk=128):
    token_splitter = SentenceTransformersTokenTextSplitter(
        chunk_overlap=chunk_overlap,
        model_name=sentence_transformer_model,
        tokens_per_chunk=tokens_per_chunk
    )
    text_chunksinTokens = []
    for text in text_chunksinChar:
        text_chunksinTokens += token_splitter.split_text(text)
    print(f"✅ {len(text_chunksinTokens)} token parçası oluşturuldu")
    return text_chunksinTokens

print("✅ PDF işleme fonksiyonları hazır!")
```

```python
# 3- ChromaDB Veritabanı Oluşturma

import chromadb
from chromadb.utils import embedding_functions

# Ayarlar
PDF_PATH = "/content/drive/MyDrive/2218121009_melike_colak.pdf"
COLLECTION_NAME = "melike_pdf"
MODEL_NAME = "distiluse-base-multilingual-cased-v1"

# Embedding fonksiyonu
embedding_function = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name=MODEL_NAME
)

# ChromaDB istemcisi oluştur
chroma_client = chromadb.PersistentClient(path="./chromadb_storage")

try:
    chroma_client.delete_collection(COLLECTION_NAME)
    print(f"⚠️ Eski '{COLLECTION_NAME}' koleksiyonu silindi")
except:
    pass

# Yeni koleksiyon oluştur
chroma_collection = chroma_client.create_collection(
    COLLECTION_NAME,
    embedding_function=embedding_function
)
print(f"✅ '{COLLECTION_NAME}' koleksiyonu oluşturuldu")

# PDF'i işle
print("\n📖 PDF işleniyor...")
pdf_texts = convert_PDF_Text(PDF_PATH)
text_chunksinChar = convert_Page_ChunkinChar(pdf_texts)
text_chunksinTokens = convert_Chunk_Token(text_chunksinChar, MODEL_NAME)

# ChromaDB'ye ekle
ids = [str(i) for i in range(len(text_chunksinTokens))]
metadatas = [{"source": "2218121009_melike_colak.pdf"} for _ in range(len(text_chunksinTokens))]

print("\n💾 Veritabanına ekleniyor...")
chroma_collection.add(
    ids=ids,
    metadatas=metadatas,
    documents=text_chunksinTokens
)

print(f"\n🎉 Başarılı! Toplam {chroma_collection.count()} parça veritabanında!")
```

```python
# 4- Belge Sorgulama ve Gemini Entegrasyonu

from google import genai
from google.genai import types
from google.colab import userdata

# Belge sorgulama
def retrieveDocs(chroma_collection, query, n_results=5):
    results = chroma_collection.query(
        query_texts=[query],
        include=["documents", "metadatas", "distances"],
        n_results=n_results
    )
    return results

# Sonuçları gösterme
def show_results(results):
    output_string = ""
    retrieved_documents = results['documents'][0]

    if len(retrieved_documents) == 0:
        return "🚩 İlgili sonuç bulunamadı..."

    for i, doc in enumerate(retrieved_documents):
        output_string += f"\n--- Parça {i+1} ---\n{doc}\n"

    return output_string

# Bağlam oluşturma
def generate_context_prompt(query, chunks):
    context_prompt = f"""### Kullanıcı Sorgusu:
{query}

### Erişilen Belgeler:
{chunks}
"""
    return context_prompt

# Gemini client'ı global olarak oluştur
GOOGLE_API_KEY = userdata.get('google_api_key')
gemini_client = genai.Client(api_key=GOOGLE_API_KEY)

# Sistem promptu
SYSTEM_PROMPT = """Sen yardımcı bir asistansın.
Kullanıcıya verilen belgelerden yararlanarak Türkçe cevap ver.
Belgelerde bilgi yoksa "Bu konuda belgede bilgi bulamadım" de."""

# Cevap üretme fonksiyonu
def generate_answer(user_prompt):
    chat_config = types.GenerateContentConfig(
        system_instruction=SYSTEM_PROMPT,
        temperature=0.7,
    )

    # v1beta için doğru model formatı
    chat = gemini_client.chats.create(
        model="models/gemini-2.5-flash",
        config=chat_config,
    )

    response = chat.send_message(user_prompt)
    return response.text

print("✅ Sorgulama ve Gemini sistemi hazır!")
```

```python
# 5-Gradio Web Arayüzü
import gradio as gr

def generateAnswerInterface(user_query, history):
    try:
        # Sorguya göre belgeleri getir
        retrieved_documents = retrieveDocs(
            chroma_collection,
            user_query,
            n_results=5
        )

        # Belgeleri formatla
        chunks = show_results(retrieved_documents)

        # Bağlam oluştur
        context_prompt = generate_context_prompt(user_query, chunks)

        # Gemini ile cevap üret
        llm_answer = generate_answer(context_prompt)

        return llm_answer

    except Exception as e:
        return f"❌ Hata oluştu: {str(e)}"

# Gradio arayüzü
demo = gr.ChatInterface(
    generateAnswerInterface,
    type="messages",
    chatbot=gr.Chatbot(height=600),
    textbox=gr.Textbox(
        placeholder="PDF'iniz hakkında bir soru sorun...",
        container=False,
        scale=7
    ),
    title="📄 PDF Soru-Cevap Asistanı",
    description="Bu bot, yüklediğiniz PDF belgesine dayalı cevaplar verir.",
    theme="soft",
    examples=[
        "Bu belge ne hakkında?",
        "Önemli noktaları özetle",
        "Hangi konular işlenmiş?"
    ],
    cache_examples=False,
)

demo.launch(share=True, debug=True)
```
